```python
import pickle 
from pathlib import Path 
import yaml
import importlib
import copy

```

## Make manifest for ResNet18 parameter decoding from SSL task heads

```python
config_dir = Path("model_configs")


equi_lmda_vals = [0.1, 0.2, 0.4, 0.5]
arch_configs = []

for lmbda in equi_lmda_vals:
    aug_in_eq_str = Path(f"model_configs/resnet18_barlow_equivariant_lmbda_1e-2_lr_2e-1_no_avgpool_eq_lmbda_{lmbda:.0e}.yaml")
    arch_configs.append(aug_in_eq_str.resolve())
    invar_aug_str = Path(f"model_configs/resnet18_barlow_equivariant_lmbda_1e-2_lr_2e-1_w_invar_augment_no_avgpool_eq_lmbda_{lmbda:.0e}.yaml")
    arch_configs.append(invar_aug_str.resolve())
 

config_dict = {ix:config for ix,config in enumerate(arch_configs)} 
manifest_dir = Path("eval_config_manifests/")
manifest_dir.mkdir(parents=True, exist_ok=True)
out_name = manifest_dir / "resnet18_barlow_equivariant_param_decoding_config_list.pkl"

with open(out_name, 'wb') as f:
    pickle.dump(config_dict, f)
```
