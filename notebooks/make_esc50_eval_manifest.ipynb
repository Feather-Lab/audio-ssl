```python
import pickle 
from pathlib import Path 
import yaml
import importlib
```

## Get set of training manifests and put into one manifest for job array

```python
train_config_manifests = list(Path("train_config_manifests").glob("*hpara_search*.pkl"))
train_config_manifests
```

```python
## unpack each manifest 
eval_configs = []
for manifest in train_config_manifests:
    with open(manifest, 'rb') as f:
        manifest = pickle.load(f)
    eval_configs.extend(list(manifest.values()))

# enumerate for job array 
eval_config_dict = {ix:config for ix,config in enumerate(eval_configs)}
print(len(eval_config_dict))
```

```python
manifest_dir = Path("eval_config_manifests/")
manifest_dir.mkdir(parents=True, exist_ok=True)
out_name = manifest_dir / "single_task_ssl_hpara_search_esc50_list_10_2024.pkl"

with open(out_name, 'wb') as f:
    pickle.dump(eval_config_dict, f)
```
