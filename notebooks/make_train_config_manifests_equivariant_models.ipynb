```python
import pickle 
from pathlib import Path 
import yaml
import importlib
import copy

```

## Make manifest for ResNet equivariant lambda sweep in new range (06/10/2025)

```python
config_dir = Path("model_configs")

base_config_path = "model_configs/resnet18_barlow_equivariant_lmbda_1e-2_lr_2e-1_w_invar_augment_no_avgpool_eq_lmbda_5e-01.yaml"

base_config = yaml.load(open(base_config_path), Loader=yaml.FullLoader)

equi_lmda_vals = [0.1, 0.2, 0.3, 0.4]

equi_arch_configs = []
for lmda in equi_lmda_vals:
    config_out_name = config_dir / f"resnet18_barlow_equivariant_lmbda_1e-2_lr_2e-1_w_invar_augment_no_avgpool_eq_lmbda_{lmda:.0e}.yaml"
    print(config_out_name)
    config = copy.deepcopy(base_config)
    config['hparas']['ssl_loss_kwargs']['lmda'] = lmda
    equi_arch_configs.append(config_out_name.resolve())

    with open(config_out_name, 'w') as f:
        yaml.dump(config, f, default_flow_style=False)

config_dict = {ix:str(config.resolve()) for ix,config in enumerate(equi_arch_configs)} 
manifest_dir = Path("train_config_manifests/")
manifest_dir.mkdir(parents=True, exist_ok=True)
out_name = manifest_dir / "resnet18_barlow_equivariant_lmbda_search_invar_to_augments.pkl"

with open(out_name, 'wb') as f:
    pickle.dump(config_dict, f)
```
