```python
import pickle 
from pathlib import Path 
import yaml
import importlib
import copy

```

### Get base equivariant model

```python
## Get list of configs to train 
config_dir = Path("model_configs/equi_lmbda_search/")
config_dir.mkdir(parents=True, exist_ok=True)

base_config_path = "model_configs/barlow_dualtask_kell2018_base_Matched_blocked_batches_lmbda_1e-2_lr_2e-1_w_augment_eq_lmbda_1e-1.yaml"
base_config_path

base_config = yaml.load(open(base_config_path), Loader=yaml.FullLoader)

equi_lmda_vals = [0.001, 0.01, 0.2, 0.3, 0.4, 0.5]


equi_arch_configs = []
for lmda in equi_lmda_vals:
    config_out_name = config_dir / f"barlow_dualtask_kell2018_base_Matched_blocked_batches_lmbda_1e-2_lr_2e-1_w_augment_eq_lmbda_{lmda:.0e}.yaml"
    print(config_out_name)
    config = copy.deepcopy(base_config)
    config['hparas']['ssl_loss_kwargs']['lmda'] = lmda
    equi_arch_configs.append(config_out_name.resolve())

    with open(config_out_name, 'w') as f:
        yaml.dump(config, f, default_flow_style=False)

```

### Save config dict for job array

```python
manifest_dir = Path("train_config_manifests/")
manifest_dir.mkdir(parents=True, exist_ok=True)
out_name = manifest_dir / "barlow_equivariant_lmbda_search_kell2018.pkl"

config_dict = {ix:config for ix,config in enumerate(equi_arch_configs)} 

with open(out_name, 'wb') as f:
    pickle.dump(config_dict, f)
    
```
